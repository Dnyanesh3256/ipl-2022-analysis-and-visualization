**Now, Here comes some Basic Questions**

1. Which team won the most matches?

```python
match_wins = df["match_winner"].value_counts()
sns.barplot(x=match_wins.values, y=match_wins.index, palette="magma")
plt.title("matches won by teams")
plt.xlabel("no. of matches won")
```

2. Toss Decision Trends

```python
sns.countplot(x=df["toss_decision"], palette="rocket")
plt.title("toss decision trends")
```

3. Toss Winner vs Match Winner

```python
count = df[df["toss_winner"] == df["match_winner"]]["match_id"].count()
perc = (count * 100) / df.shape[0]
perc.round(2)
```

4. How do teams win? (Runs vs Wickets)

```python
sns.countplot(x=df["won_by"], palette="magma")
plt.title("TEAMS WON BY")
```

###  Key Player Performances

1. Most "Player of the Match" Awards

```python
count = df["player_of_the_match"].value_counts().head(10)
sns.barplot(x=count.values, y=count.index, palette="mako")
plt.title("Most Player of the Match Awards")
```

2. Top Scorers

```python
high_scorers = df.groupby("top_scorer")["highscore"].sum().sort_values(ascending=False).head(10)
high_scorers.plot(kind="barh")
plt.title("Top Scorers")
```

3. 10 Best Bowling Figures

```python
df["highest_wickets"] = df["best_bowling_figure"].apply(lambda x: x.split("--")[0])
df["highest_wickets"] = df["highest_wickets"].astype(int)
top_bowlers = df.groupby("best_bowling")["highest_wickets"].sum().sort_values(ascending=False).head(10)
top_bowlers.plot(kind="barh")
plt.title("10 Best Bowling Figures")
```

###  Venue Analysis

Most Matches Played by Venue

```python
venue_count = df["venue"].value_counts()
sns.barplot(x=venue_count.values, y=venue_count.index, palette="magma")
```
